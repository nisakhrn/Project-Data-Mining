# Clustering Kabupaten/Kota di Aceh Berdasarkan Profil Pendidikan (K-Means)


## 1. Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

sns.set_style("whitegrid")

pd.set_option("display.max_columns",None)

print("Library berhasil diimport.")
```

## 2. Load Dataset

```python

df = pd.read_csv('data_bersih.csv')
print("Ukuran data:", df.shape)
print("Jumlah kabupaten/kota:", df['kabupaten_kota'].nunique())
print("Rentang tahun:", df['tahun'].min(), "-", df['tahun'].max())
print("Missing value:", df.isnull().sum().sum())
print("Duplikat:", df.duplicated().sum())
df.head()
```

## 3. Agregasi ke Level Kabupaten/Kota

Data awal berbentuk panel (kabupaten × tahun). Karena tujuan analisis adalah mengelompokkan **kabupaten/kota**, data diagregasi dengan mengambil **rata-rata tahun 2016–2025** per kabupaten/kota. Pendekatan ini merepresentasikan kondisi umum kabupaten selama satu dekade dan mengurangi noise akibat fluktuasi tahunan.

```python
numeric_cols = df.columns.drop([
    "kabupaten_kota",
    "tahun"
])

agg = (
    df
    .groupby("kabupaten_kota")[numeric_cols]
    .mean()
    .reset_index()
)

print(agg.shape)

display(agg.head())
```

## 4. Feature Engineering

## 4.1. Integrasi Fitur Spasial untuk Spatial K-Means
**Penjelasan Metodologi Analisis Spasial:**
Untuk menjawab kebutuhan **Analisis Spasial (Spatial Analysis)**, algoritma K-Means kini menggunakan matriks fitur terintegrasi yang mencakup **fitur geografis/spasial** (`latitude` dan `longitude`) dari koordinat sentroid masing-masing Kabupaten/Kota di Provinsi Aceh, disamping indikator atribut pendidikan.

Dengan pendekatan **Spatial-Attribute K-Means** ini, pengelompokan yang dihasilkan oleh K-Means memperhitungkan dua dimensi secara simultan:
1. **Kedekatan Lokasi Geografis (Spatial Proximity)**: Memastikan wilayah yang secara geografis berdekatan/bertetangga dipertimbangkan dalam pembentukan klaster.
2. **Kemiripan Indikator Pendidikan (Attribute Similarity)**: Memastikan karakteristik atribut pendidikan (seperti HLS, RLS, AMH, dan rasio-rasio pendidikan) tetap dioptimalkan.


```python

levels = ["sd", "smp", "sma", "smk"]

feat = pd.DataFrame()

# Identitas
feat["kabupaten_kota"] = agg["kabupaten_kota"]

# Fitur Koordinat Spasial (Latitude & Longitude)
koordinat_aceh = {
    'Kabupaten Aceh Barat': (4.45, 96.16),
    'Kabupaten Aceh Barat Daya': (3.83, 96.89),
    'Kabupaten Aceh Besar': (5.38, 95.52),
    'Kabupaten Aceh Jaya': (4.72, 95.64),
    'Kabupaten Aceh Selatan': (3.31, 97.18),
    'Kabupaten Aceh Singkil': (2.42, 97.93),
    'Kabupaten Aceh Tengah': (4.54, 96.91),
    'Kabupaten Aceh Tenggara': (3.37, 97.80),
    'Kabupaten Aceh Timur': (4.63, 97.63),
    'Kabupaten Aceh Utara': (4.97, 97.14),
    'Kabupaten Bener Meriah': (4.73, 96.86),
    'Kabupaten Bireuen': (5.14, 96.62),
    'Kabupaten Gayo Lues': (3.96, 97.33),
    'Kabupaten Nagan Raya': (4.16, 96.43),
    'Kabupaten Pidie': (5.16, 95.96),
    'Kabupaten Pidie Jaya': (5.23, 96.22),
    'Kabupaten Simeulue': (2.61, 96.08),
    'Kota Banda Aceh': (5.55, 95.32),
    'Kota Langsa': (4.47, 97.96),
    'Kota Lhokseumawe': (5.18, 97.14),
    'Kota Sabang': (5.88, 95.32),
    'Kota Subulussalam': (2.64, 97.93),
    'Kabupaten Aceh Tamiang': (4.25, 98.02)
}
feat['latitude'] = feat['kabupaten_kota'].map(lambda x: koordinat_aceh.get(x, (0,0))[0])
feat['longitude'] = feat['kabupaten_kota'].map(lambda x: koordinat_aceh.get(x, (0,0))[1])

# Indikator pendidikan
feat["hls"] = agg["hls__nilai"]
feat["rls"] = agg["rls__nilai"]
feat["amh"] = agg["amh__nilai"]

for lv in levels:

    guru = agg[f"{lv}__Jumlah Guru {lv.upper()} (Negeri+Swasta)"]
    murid = agg[f"{lv}__Jumlah Murid {lv.upper()} (Negeri+Swasta)"]
    sekolah = agg[f"{lv}__Jumlah Sekolah {lv.upper()} (Negeri+Swasta)"]

    # Rasio Murid per Guru
    feat[f"{lv}_rasio_murid_guru"] = murid / guru

    # Rasio Murid per Sekolah
    feat[f"{lv}_rasio_murid_sekolah"] = murid / sekolah

# Mengatasi nilai kosong jika ada
feat = feat.replace([np.inf, -np.inf], np.nan)
feat = feat.fillna(0)

print("Ukuran fitur :", feat.shape)

display(feat.head())
```

## 5. Standarisasi Fitur


```python
X = feat.drop(columns=["kabupaten_kota"])

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Ukuran data:", X_scaled.shape)
print("Mean :", np.round(X_scaled.mean(axis=0),3)[:5])
print("Std  :", np.round(X_scaled.std(axis=0),3)[:5])
```

## 6. Exploratory Data Analysis - Korelasi Antar Fitur

```python

plt.figure(figsize=(11, 9))
corr = feat.drop(columns=['kabupaten_kota']).corr()
sns.heatmap(corr, cmap='RdBu_r', center=0, annot=True, fmt='.2f', linewidths=0.3, annot_kws={'size':7})
plt.title('Korelasi Antar Fitur')
plt.tight_layout()
plt.show()
```

## 7. Menentukan Jumlah Cluster Optimal (Elbow Method & Silhouette Score)

```python
K_range = range(2, 9)
inertias = []
sil_scores = []

for k in K_range:
    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = km.fit_predict(X_scaled)

    inertias.append(km.inertia_)
    sil_scores.append(
        silhouette_score(X_scaled, labels)
    )

fig, axes = plt.subplots(1,2,figsize=(13,4.5))

axes[0].plot(K_range,inertias,marker='o')
axes[0].set_title("Elbow Method")
axes[0].set_xlabel("Jumlah Cluster (k)")
axes[0].set_ylabel("Inertia")

axes[1].plot(K_range,sil_scores,marker='o', color='#e6550d')
axes[1].set_title("Silhouette Score")
axes[1].set_xlabel("Jumlah Cluster (k)")
axes[1].set_ylabel("Silhouette Score")

plt.tight_layout()
plt.show()

for k,i,s in zip(K_range,inertias,sil_scores):
    print(f"k={k}: inertia={i:.2f}, silhouette={s:.4f}")

best_k = K_range[np.argmax(sil_scores)]

print("="*40)
print("Jumlah cluster terbaik :", best_k)
print("Silhouette terbaik     :", round(max(sil_scores),4))
```

## 8. Fit K-Means

```python
best_k = K_range[np.argmax(sil_scores)]

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

feat['cluster'] = kmeans.fit_predict(X_scaled)

print("Jumlah anggota tiap cluster")
print(feat['cluster'].value_counts().sort_index())

print()

for c in sorted(feat['cluster'].unique()):
    anggota = feat.loc[
        feat['cluster']==c,
        'kabupaten_kota'
    ].tolist()

    print(f"Cluster {c}")
    print(", ".join(anggota))
    print()
```

## 9. Visualisasi Cluster dengan PCA

```python

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
print("Variance explained oleh 2 komponen PCA:", np.round(pca.explained_variance_ratio_, 3), 
      "-> total", round(pca.explained_variance_ratio_.sum()*100, 1), "%")

plt.figure(figsize=(9, 7))
palette = ['#2c7fb8', '#e6550d', '#31a354']
for c in sorted(feat['cluster'].unique()):
    mask = feat['cluster'] == c
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1], s=110, color=palette[c], label=f'Cluster {c}', edgecolor='white')

for i, row in feat.iterrows():
    plt.annotate(row['kabupaten_kota'].replace('Kabupaten ', '').replace('Kota ', ''),
                 (X_pca[i, 0], X_pca[i, 1]), fontsize=8, xytext=(4, 4), textcoords='offset points')

plt.xlabel('PCA Komponen 1')
plt.ylabel('PCA Komponen 2')
plt.title('Visualisasi Cluster Kabupaten/Kota Aceh (PCA 2D)')
plt.legend()
plt.tight_layout()
plt.show()
```

## 10. Profil Rata-Rata Tiap Cluster

```python
profile = (
    feat
    .drop(columns='kabupaten_kota')
    .groupby('cluster')
    .mean(numeric_only=True)
    .round(2)
)

display(profile)

print()

print("Jumlah anggota cluster")

print(feat['cluster'].value_counts().sort_index())
```

```python

fitur_utama = ['hls', 'rls', 'amh']
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, f in zip(axes, fitur_utama):
    ax.bar(profile.index.astype(str), profile[f], color=['#2c7fb8', '#e6550d', '#31a354'])
    ax.set_title(f.upper())
    ax.set_xlabel('Cluster')
plt.tight_layout()
plt.show()
```

```python
profile.T.plot(
    figsize=(13,7),
    marker="o"
)

plt.title("Perbandingan Rata-rata Fitur Tiap Cluster")
plt.ylabel("Nilai Rata-rata")
plt.xlabel("Fitur")
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
hasil_cluster = feat.copy()

hasil_cluster.to_csv(
    "hasil_cluster_tema1.csv",
    index=False
)

print("Hasil clustering berhasil disimpan.")
```

## 11. Kesimpulan Analisis

Berdasarkan hasil K-Means (k=2), kabupaten/kota di Aceh terbagi menjadi dua kelompok utama:

- **Cluster 0 (19 kabupaten/kota)** mewakili mayoritas wilayah kabupaten dengan karakteristik layanan pendidikan yang relatif lebih longgar. Secara umum, rasio murid per guru dan murid per sekolah lebih rendah dibanding Cluster 1, sehingga tekanan kapasitas layanan cenderung lebih ringan.

- **Cluster 1 (4 wilayah: Kabupaten Aceh Tamiang, Kota Banda Aceh, Kota Langsa, dan Kota Lhokseumawe)** menunjukkan profil wilayah yang lebih urban dengan capaian pendidikan lebih tinggi (terutama pada HLS dan AMH), namun sekaligus memiliki rasio murid per guru/sekolah yang lebih tinggi. Ini mengindikasikan beban layanan pendidikan yang lebih padat.

**Interpretasi utama:**
Cluster 1 dapat dibaca sebagai kelompok wilayah dengan capaian relatif tinggi tetapi tekanan layanan juga tinggi, sedangkan Cluster 0 adalah kelompok mayoritas wilayah dengan capaian yang lebih moderat dan kepadatan layanan yang lebih rendah.
